```python
import pandas as pd

file_path = "xy_data.csv"
df = pd.read_csv(file_path)

df.info(), df.head()
```

```python
import numpy as np
from scipy.optimize import least_squares
import matplotlib.pyplot as plt

x_data = df["x"].values
y_data = df["y"].values

t = np.linspace(6, 60, len(x_data))

def model(params, t):
    theta, M, X = params
    theta_rad = np.deg2rad(theta)
    x_model = (t * np.cos(theta_rad) - np.exp(M * np.abs(t)) * np.sin(0.3 * t) * np.sin(theta_rad) + X)
    y_model = (42 + t * np.sin(theta_rad) + np.exp(M * np.abs(t)) * np.sin(0.3 * t) * np.cos(theta_rad))
    return x_model, y_model

def residuals(params, t, x_data, y_data):
    x_model, y_model = model(params, t)
    return np.concatenate([(x_model - x_data), (y_model - y_data)])

initial_guess = [25, 0.0, 50]
bounds = ([0, -0.05, 0], [50, 0.05, 100])

result = least_squares(residuals, initial_guess, bounds=bounds, args=(t, x_data, y_data))

theta_opt, M_opt, X_opt = result.x

x_fit, y_fit = model(result.x, t)

theta_opt, M_opt, X_opt
```
